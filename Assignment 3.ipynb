```python
import numpy as np
from scipy.integrate import odeint, solve_ivp, quad
import matplotlib.pyplot as plt
import time
```

```python
def Koala(t,X0,betaprime_mf,betaprime_fm,r,alpha,mu):
    N,Sf,If,Sm,Im = X0
    Ndot = -r*N**2+2*r*(Sf+alpha*If)-mu*(If+Im)
    Sfdot = r*(Sf+alpha*If)-betaprime_fm*Sf*Im/N-r*N*Sf
    Ifdot = betaprime_fm*Sf*Im/N-(r*N+mu)*If
    Smdot = r*(Sf+alpha*If)-betaprime_mf*Sm*If/N-r*N*Sm
    Imdot = betaprime_mf*Sm*If/N-(r*N+mu)*Im
    return [Ndot,Sfdot,Ifdot,Smdot,Imdot]
```

```python
vals2 = np.arange(0,1,0.01)
extinct_alpha = []
extinct_mu = []
survived_alpha = []
survived_mu = []
```

```python
for mu in vals2:
    for alpha in vals2:
        soln = solve_ivp(Koala,t,X0,args=(betaprime_mf,betaprime_fm,r,alpha,mu))
        Y = soln.y.T[:,0]
        if Y[-1]<0.025:
            extinct_alpha.append(alpha)
            extinct_mu.append(mu)
        else:
            survived_alpha.append(alpha)
            survived_mu.append(mu)
plt.plot(extinct_alpha,extinct_mu,'r.')
plt.plot(survived_alpha,survived_mu,'b.')
```
